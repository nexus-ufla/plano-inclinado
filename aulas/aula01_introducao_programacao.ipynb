## 7. Funções

Funções agrupam instruções reutilizáveis. Elas ajudam a organizar o código e evitam repetição.

A estrutura básica é:


Outro exemplo: posição de um corpo em movimento retilineo uniforme.

## 9. Laços de repetição

Laços permitem repetir comandos. Eles são fundamentais em simulações, porque normalmente calculamos a evolução de um sistema passo a passo no tempo.

````python
for i in range(5):
    print("passo", i)
````

Exemplo: atualizar a posição de um corpo com velocidade constante a cada intervalo de tempo `dt`.

````python
posicao = 0.0
velocidade = 2.0
dt = 0.5

for passo in range(6):
    print("passo:", passo, "posição:", posicao)
    posicao = posicao + velocidade * dt
````

## 12. Matplotlib

Matplotlib é uma biblioteca para criar gráficos. Ela é muito usada em projetos científicos.

Por convenção, importamos assim:

````python
import matplotlib.pyplot as plt
````

### Gráfico simples

````python
tempo = np.linspace(0, 5, 100)
velocidade = 3.0
posicao = velocidade * tempo

plt.figure(figsize=(7, 4))
plt.plot(tempo, posicao)
plt.xlabel("tempo (s)")
plt.ylabel("posição (m)")
plt.title("Movimento retilineo uniforme")
plt.grid(True)
plt.show()
````

## 13. Pandas

Pandas é uma biblioteca para trabalhar com dados em formato de tabela. Ela é muito útil quando temos dados experimentais, resultados de simulações ou medições organizadas em colunas.

A estrutura principal do Pandas é o `DataFrame`, que pode ser entendido como uma tabela com linhas e colunas.

Por convenção, importamos Pandas assim:

````python
import pandas as pd
````

### Criando uma tabela simples

Vamos criar uma tabela com tempo, posição e velocidade de um corpo em movimento retilíneo uniforme.

````python
tempo = np.array([0, 1, 2, 3, 4, 5])
velocidade = 2.0
posicao = velocidade * tempo

dados = pd.DataFrame({
    "tempo (s)": tempo,
    "posição (m)": posicao,
    "velocidade (m/s)": velocidade
})

dados
````

### Acessando colunas

Podemos selecionar uma coluna pelo nome. Isso é útil para calcular estatísticas ou gerar gráficos a partir dos dados.

````python
print(dados["tempo (s)"])
print(dados["posição (m)"])
````

### Estatísticas simples

Pandas facilita cálculos como média, valor mínimo e valor máximo.

````python
print("posição média:", dados["posição (m)"].mean())
print("posição m?xima:", dados["posição (m)"].max())
print("posição m?nima:", dados["posição (m)"].min())
````

### Gráfico usando dados de uma tabela

Também podemos usar as colunas do `DataFrame` diretamente no Matplotlib.

````python
plt.figure(figsize=(7, 4))
plt.plot(dados["tempo (s)"], dados["posição (m)"], marker="o")
plt.xlabel("tempo (s)")
plt.ylabel("posição (m)")
plt.title("Dados organizados com Pandas")
plt.grid(True)
plt.show()
````

## 14. SciPy

SciPy é uma biblioteca com ferramentas científicas mais avançadas. Ela inclui recursos para integração, interpolação, otimização, estatística e muito mais.

Nesta primeira aula, vamos ver apenas uma ideia simples: usar uma função de integração numérica.

````python
from scipy.integrate import quad

def f(x):
    return x ** 2

area, erro = quad(f, 0, 1)

print("integral de x^2 entre 0 e 1:", area)
print("estimativa de erro:", erro)
````

O resultado esperado e apróximadamente:

$$\int_0^1 x^2 dx = \frac{1}{3}$$

## 15. Miniaplicação: movimento com aceleração constante

Agora vamos juntar variáveis, funções, NumPy e Matplotlib.

Para um corpo com aceleração constante:

$$x(t) = x_0 + v_0t + \frac{1}{2}at^2$$

Essa forma vai aparecer novamente na aula do plano inclinado.

````python
def posicao_acelerada(t, x0, v0, a):
    return x0 + v0 * t + 0.5 * a * t ** 2

tempo = np.linspace(0, 5, 100)
x0 = 0.0
v0 = 0.0
a = 2.0

x = posicao_acelerada(tempo, x0, v0, a)

plt.figure(figsize=(7, 4))
plt.plot(tempo, x, label="x(t)")
plt.xlabel("tempo (s)")
plt.ylabel("posição (m)")
plt.title("Movimento com aceleração constante")
plt.legend()
plt.grid(True)
plt.show()
````

## 16. Ponte para o plano inclinado

No plano inclinado sem atrito, a componente da gravidade ao longo do plano gera uma aceleração:

$$a = g\sin(\theta)$$

Na aula 2, vamos usar essa aceleração para atualizar posição e velocidade passo a passo pelo método de Euler.

Por enquanto, vamos apenas calcular essa aceleração para alguns ângulos.

````python
g = 9.81
angulos_graus = np.array([10, 20, 30, 40, 50])
angulos_rad = np.deg2rad(angulos_graus)

aceleracoes = g * np.sin(angulos_rad)

for angulo, aceleracao in zip(angulos_graus, aceleracoes):
    print(f"theta = {angulo} graus -> a = {aceleracao:.2f} m/s^2")
````

````python
plt.figure(figsize=(7, 4))
plt.plot(angulos_graus, aceleracoes, marker="o")
plt.xlabel("ângulo (graus)")
plt.ylabel("aceleração (m/s^2)")
plt.title("Aceleracao em um plano inclinado sem atrito")
plt.grid(True)
plt.show()
````
